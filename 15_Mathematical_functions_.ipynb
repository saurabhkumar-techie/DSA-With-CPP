```python
import numpy as np
```

```python
arr = np.array([4,9,16,25,36])

print(np.sqrt(arr))
```

```python
#absolute value
arr = np.array([-10,-5,0,5,10])

print(np.abs(arr))
```

```python
#rounding
arr = np.array([2.345,3.678,4.555])

print(np.round(arr,2))
print(np.round(arr,1))
print(np.round(arr))
```

```python
#floor and ceil
arr = np.array([2.7,3.2,-2.3])
print(np.floor(arr)) #niche ki taraf nearest inetger
print(np.ceil(arr)) # upar ki taraf nearest integer
```

```python
#exponential
arr = np.array([0,1,2])
print(np.exp(arr))
```

```python
#natural logarithm
arr = np.array([1, np.e, np.e ** 2])
print(np.log(arr))
```

```python
#base-10 logarithm
arr = np.array([1, 10, 100, 1000])

print(np.log10(arr))
```

```python
#Trigonometric functions (angles into radians)
angles = np.array([0, np.pi / 2, np.pi])
print(np.round(np.sin(angles),2))
print(np.round(np.cos(angles),2))
```
