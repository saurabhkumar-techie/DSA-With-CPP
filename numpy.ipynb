# 01 - Introduction to NumPy

## 1. What is NumPy?

NumPy stands for Numerical Python.

It is a Python library used for numerical computing.
It provides multidimensional arrays and efficient
mathematical operations.

## 2. Why Do We Use NumPy?

- Fast numerical calculations
- Multidimensional array handling
- Mathematical and statistical operations
- Linear algebra
- Data Science and Machine Learning

## 3. Main Object

The main object in NumPy is ndarray,
which stands for N-dimensional array.

## 4. Applications

- Data Science
- Machine Learning
- Artificial Intelligence
- Scientific Computing
- Statistical Analysis
